```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset.xlsx")

df.head()
```

## Numerical Statistics

```python
numerical_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

df[numerical_columns].describe()
```

## Categorical columns

```python
print(df["Gender"].value_counts())
print(df["City"].value_counts())
print(df["Product"].value_counts())
print(df["Category"].value_counts())
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Age"].dropna(), bins=10, edgecolor="black")

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

## Total Sales distribution

```python
plt.figure(figsize=(8,5))

plt.hist(df["Total_Sales"], bins=20, edgecolor="black")

plt.title("Total Sales Distribution")
plt.xlabel("Total Sales")
plt.ylabel("Number of Orders")

plt.show()
```

## Gender bar chart

```python
plt.figure(figsize=(6,4))

df["Gender"].value_counts().plot(kind="bar")

plt.title("Orders by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Orders")

plt.show()
```

## Category bar chart

```python
plt.figure(figsize=(7,4))

df["Category"].value_counts().plot(kind="bar")

plt.title("Orders by Category")
plt.xlabel("Category")
plt.ylabel("Number of Orders")

plt.show()
```

## Product bar chart

```python
plt.figure(figsize=(8,5))

df["Product"].value_counts().plot(kind="bar")

plt.title("Orders by Product")
plt.xlabel("Product")
plt.ylabel("Number of Orders")

plt.show()
```

## City distribution

```python
plt.figure(figsize=(8,5))

df["City"].value_counts().plot(kind="bar")

plt.title("Orders by City")
plt.xlabel("City")
plt.ylabel("Number of Orders")

plt.show()
```

## Scatter plot

Age vs Total Sales

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Age"], df["Total_Sales"])

plt.title("Age vs Total Sales")
plt.xlabel("Age")
plt.ylabel("Total Sales")

plt.show()
```

## Quantity vs Total Sales

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Quantity"], df["Total_Sales"])

plt.title("Quantity vs Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")

plt.show()
```

## Unit Price vs Total Sales

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Unit_Price"], df["Total_Sales"])

plt.title("Unit Price vs Total Sales")
plt.xlabel("Unit Price")
plt.ylabel("Total Sales")

plt.show()
```

## Correlation Heatmap

Sirf numerical columns lenge:

```python
numeric_df = df[[
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]]
correlation = numeric_df.corr()

correlation
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

# Pair Plot

```python
sns.pairplot(
    df[[
        "Age",
        "Quantity",
        "Unit_Price",
        "Total_Sales"
    ]].dropna()
)

plt.show()
```

## Category vs Multivariate Analysis

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Category",
    y="Age"
)

plt.title("Age Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Age")

plt.xticks(rotation=45)

plt.show()
```

## Category vs Total Sales

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Category",
    y="Total_Sales"
)

plt.title("Total Sales Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.show()
```
